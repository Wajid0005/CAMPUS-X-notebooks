#**1. Lowercasing**
Convert every character in your text to lowercase. This ensures words like "Movie", "MOVIE", and "movie" are treated as the exact same word instead of three different ones.

```python
df = df.astype(str)
```

```python
df['headlines'] = df['headlines'].str.lower()
df['description'] = df['description'].str.lower()
df['content'] = df['content'].str.lower()
df['url'] = df['url'].str.lower()
df['category'] = df['category'].str.lower()
```

#**2. Removal of HTML Tags & URLs**
Web-scraped data often contains leftover code (like <br>, <p>) and web links (http://...). Strip all of these out because they add noise without giving any real meaning to the text.

```python
print(df.url.head(1))
```

```python
import re

# Regex to extract the main domain (e.g., example.com from http://www.example.com/path)
# It handles http/https and optional www. prefixes.
# The capturing group ([^/]+) will get everything between the protocol/www and the next slash or end of string.
url_pattern = r'(?:https?://)?(?:www\.)?([^/]+)'

# Apply the regex to the 'url' column and assign the extracted domain back to the 'url' column
df['url'] = df['url'].str.extract(url_pattern)

# Print the head of the updated url column to show the changes
print(df['url'].head())
```

# **3. Removal of Punctuation & Special Characters**
Delete all commas, full stops, exclamation marks, and symbols (#, @, %, numbers). Keep only alphabetic letters and plain spaces.

```python
in
```

```python
import re

# Define a function to remove punctuation and numbers, keeping only alphabetic characters and spaces
def remove_punctuation_numbers(text):
    # Remove numbers
    text = re.sub(r'\d+', '', text)
    # Remove punctuation and special characters (keep only letters and spaces)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    return text

# Apply the function to the relevant text columns
df['headlines'] = df['headlines'].apply(remove_punctuation_numbers)
df['description'] = df['description'].apply(remove_punctuation_numbers)
df['content'] = df['content'].apply(remove_punctuation_numbers)

# Display the head of the updated DataFrame to show the changes in the cleaned columns
print(df[['headlines', 'description', 'content']].head())
```

# **4. Tokenization**
Break down the continuous string of text into individual words or "tokens." For example, "I love movies" becomes ['i', 'love', 'movies'].

```python
import nltk
from nltk.tokenize import word_tokenize

# Ensure the 'punkt_tab' tokenizer is downloaded. It's idempotent, so it won't re-download if already present.
nltk.download('punkt_tab', quiet=True)

# Define a function to tokenize text, handling potential non-string inputs
def tokenize_text(text):
    if isinstance(text, str):
        return word_tokenize(text)
    return []

# Apply tokenization to the relevant text columns
df['headlines_tokens'] = df['headlines'].apply(tokenize_text)
df['description_tokens'] = df['description'].apply(tokenize_text)
df['content_tokens'] = df['content'].apply(tokenize_text)

# Display the head of the updated DataFrame with new tokenized columns
print(df[['headlines', 'headlines_tokens', 'description', 'description_tokens', 'content', 'content_tokens']].head())
```

#**5. Removal of Stopwords**
Filter out high-frequency grammar words that carry almost no semantic value for NLP models—such as "is", "the", "at", "and", or "which".

```python
import nltk
from nltk.corpus import stopwords

# Download the 'stopwords' corpus if not already downloaded. It's idempotent.
nltk.download('stopwords', quiet=True)

# Get the English stopwords list
stop_words = set(stopwords.words('english'))

# Define a function to remove stopwords from a list of tokens
def remove_stopwords(tokens):
    if isinstance(tokens, list):
        return [word for word in tokens if word not in stop_words]
    return tokens

# Apply the function to the tokenized columns
df['headlines_tokens'] = df['headlines_tokens'].apply(remove_stopwords)
df['description_tokens'] = df['description_tokens'].apply(remove_stopwords)
df['content_tokens'] = df['content_tokens'].apply(remove_stopwords)

print("Stopwords successfully removed from 'headlines_tokens', 'description_tokens', and 'content_tokens'.")
print("\nHere's a peek at the updated tokenized columns:")
print(df[['headlines_tokens', 'description_tokens', 'content_tokens']].head())
```

```python
columns_to_drop = ['headlines', 'description', 'content']
df = df.drop(columns=columns_to_drop)

# Define the new order of columns: tokenized columns first, then others
tokenized_cols = ['headlines_tokens', 'description_tokens', 'content_tokens']
other_cols = [col for col in df.columns if col not in tokenized_cols]

# Combine and reorder
new_column_order = tokenized_cols + other_cols
df = df[new_column_order]

print("Original 'headlines', 'description', and 'content' columns removed, and tokenized columns reordered to the front.")
print("\nHere's the head of your updated DataFrame:")
print(df.head())
```

#**6. Text Normalization (Stemming or Lemmatization)**
Reduce remaining words to their base or root form so your model recognizes different variations of a single word:

Stemming: Chops off word endings coarsely (e.g., "running", "ran", "runs" → "run").

Lemmatization: Uses a dictionary lookup to resolve words to their true linguistic base (e.g., "better" → "good").

```python
import nltk
from nltk.stem import WordNetLemmatizer

# Download necessary NLTK data for lemmatization
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

# Initialize the WordNet Lemmatizer
lemmatizer = WordNetLemmatizer()

# Define a function to lemmatize a list of tokens
def lemmatize_tokens(tokens):
    if isinstance(tokens, list):
        return [lemmatizer.lemmatize(word) for word in tokens]
    return tokens

# Apply lemmatization to the tokenized columns
df['headlines_tokens'] = df['headlines_tokens'].apply(lemmatize_tokens)
df['description_tokens'] = df['description_tokens'].apply(lemmatize_tokens)
df['content_tokens'] = df['content_tokens'].apply(lemmatize_tokens)

print("Text normalization (lemmatization) applied to 'headlines_tokens', 'description_tokens', and 'content_tokens'.")
print("\nHere's a peek at the updated tokenized columns after lemmatization:")
print(df[['headlines_tokens', 'description_tokens', 'content_tokens']].head())
```
